### Today’s big idea
A `for` loop is great when we know what items we want to visit.

A `while` loop is great when we want to say:

> Keep going while this condition is true.

## 1. Getting User Input with `input()`

`input()` pauses the program and waits for the user to type something.

```python
name = input("What is your name? ")
print(f"Hi, {name}!")
```

Think of `input()` as Python handing the microphone to the user. 🎤

````python
name = input("What is your name? ")
print(f"Hi, {name}! Nice to meet you.")
````

### Write clear prompts

Compare:

```python
age = input("Enter: ")
```

with:

```python
age = input("How old are you? ")
```

Both work. Only one is friendly. 😄

## 2. `input()` Gives Us Text

Important rule:

> `input()` returns a string.

Even if the user types `25`, Python initially receives `"25"` as text.

````python
age = input("How old are you? ")
print(age)
print(type(age))
````

````python
age = int(input("How old are you? "))

if age >= 18:
    print("You are an adult.")
else:
    print("You are under 18.")
````

## 3. A Useful Number Tool: `%` Modulo

`%` gives the remainder after division.

```python
10 % 3   # 1
12 % 3   # 0
```

Think: 10 cookies packed into bags of 3 → 1 cookie is left. 🍪

A common use is checking even vs. odd.

````python
number = int(input("Enter a whole number: "))

if number % 2 == 0:
    print("That number is even.")
else:
    print("That number is odd.")
````

## 4. Basic `while` Loops

A `while` loop repeats as long as its condition is `True`.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

A useful mental model:

> Start → Check → Run → Update → Check again

````python
count = 1

while count <= 5:
    print(f"Count: {count}")
    count += 1

print("Done!")
````

### ⚠️ Infinite loops

What is wrong here?

```python
count = 1

while count <= 5:
    print(count)
```

`count` never changes, so the condition never becomes false.

Python is not being stubborn—it is following our instructions perfectly. 😄

If a loop runs forever, ask:
1. What makes it continue?
2. What eventually makes that condition false?

## 5. Letting the User Choose When to Quit

Sometimes we do not know whether the user will repeat something 3 times or 30 times.

That is a perfect job for `while`.

```python
message = ""

while message != "quit":
    message = input("Type something, or 'quit' to stop: ")
```

`quit` acts like an exit sign for the loop. 🚪

````python
message = ""

while message.lower() != "quit":
    message = input("Say something (or type 'quit' to stop): ")

    if message.lower() != "quit":
        print(f"You said: {message}")

print("Goodbye!")
````

## 6. More Ways to Control a Loop

### A. A flag

A flag is just a Boolean ON/OFF switch.

```python
active = True

while active:
    ...
```

When something changes `active` to `False`, the loop stops.

````python
active = True

while active:
    command = input("Enter a command ('quit' to stop): ")

    if command.lower() == "quit":
        active = False
    else:
        print(f"Running: {command}")

print("Program stopped.")
````

### B. `break`

`break` exits the loop immediately.

```python
while True:
    city = input("Enter a city, or 'quit': ")

    if city == "quit":
        break
```

`while True` keeps the door open; `break` is the exit door.

### C. `continue`

`continue` skips the rest of the current round and starts the next one.

- `break` → leave the whole loop
- `continue` → skip this round

````python
number = 0

while number < 10:
    number += 1

    if number % 2 == 0:
        continue

    print(number)
````

## 7. Using `while` with Lists and Dictionaries

A `while` loop is useful when a collection changes while the program runs.

### Move items from one list to another

````python
pending_orders = ["latte", "tea", "smoothie"]
completed_orders = []

while pending_orders:
    current_order = pending_orders.pop(0)
    print(f"Preparing: {current_order}")
    completed_orders.append(current_order)

print("Completed:", completed_orders)
````

### Build a dictionary while the program runs

Now we can combine `input()`, `while`, `if`, and dictionaries.

````python
responses = {}
active = True

while active:
    name = input("Your name: ")
    language = input("Favorite programming language: ")
    responses[name] = language

    again = input("Add another response? (yes/no): ")
    if again.lower() == "no":
        active = False

print("\nSurvey results:")
for name, language in responses.items():
    print(f"{name}: {language}")
````
