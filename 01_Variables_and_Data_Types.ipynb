## A quick Python origin story

Python was created by **Guido van Rossum**, a Dutch programmer working at CWI in the Netherlands.

During the **1989 Christmas holidays**, Guido had extra time and began experimenting with a new scripting language. He wanted something that kept useful ideas from the ABC language, could work with the Amoeba operating system, and was general enough to grow beyond one project. Python continued developing through 1990 and was first shared publicly on Usenet in **February 1991**.

### Why is it called “Python”? 🐍

Surprisingly, the language was **not named after the snake**. Guido was reading scripts from the BBC comedy series *Monty Python's Flying Circus*. He wanted a name that was short, unique, and a little mysterious, so he chose **Python**.

> Fun connection: Python was born from curiosity and a holiday side project. You do not need a perfect master plan before you begin building something.

Source: [Python Software Foundation — General Python FAQ](https://docs.python.org/3/faq/general.html#why-was-python-created-in-the-first-place)

## 2. Variables: names for values

A variable connects a **name** to a **value**.


```text
course_name  ─────▶  "Intro to Python"
   label                  value
```

The equals sign `=` means **assignment**: take the value on the right and give it the name on the left.



````python
course_name = "Intro to Python"
room_number = 204

print(course_name)
print(room_number)
````

### Variables can change

Python keeps the **most recently assigned value**.



````python
status = "Learning"
print(status)

status = "Still learning, but more confident!"
print(status)
````

### Variable naming rules

✅ Good names:

- Use letters, numbers, and underscores: `student_name`, `score_1`
- Start with a letter or underscore, not a number
- Use lowercase `snake_case` for ordinary variables
- Choose short but descriptive names

❌ Avoid:

- Spaces: `student name`
- Starting with a number: `1st_score`
- Python keywords or built-in names: `class`, `print`
- Confusing names such as lowercase `l` or uppercase `O`

**Good code is written for humans first.**


````python
student_name = "Maya"
quiz_score = 9
favorite_snack = "popcorn"

print(student_name, quiz_score, favorite_snack)
````

## 4. Data types: different kinds of information

Python treats different kinds of values differently.

| Type | Meaning | Example |
|---|---|---|
| `str` | text
| `int` | whole number
| `float` | decimal number
| `bool` | true or false

Use `type()` to ask Python what kind of value it sees.



````python
student = "Jordan"
completed_lessons = 3
average_score = 92.5
is_enrolled = True

print(type(student))
print(type(completed_lessons))
print(type(average_score))
print(type(is_enrolled))
````

### Same appearance, different type

`"8"` is text. `8` is a number. The quotation marks matter.



````python
print("8" + "8")   # joins text
print(8 + 8)       # adds numbers
````

## 5. Strings: text you can transform

A string is a sequence of characters inside single or double quotes.



````python
movie = "hidden figures"

print(movie.title())
print(movie.upper())
print(movie.lower())
````

### f-strings: build messages with variables

Put `f` before the opening quote and place variable names inside `{ }`.



````python
first_name = "grace"
last_name = "hopper"
language = "Python"

full_name = f"{first_name} {last_name}"
print(f"Hello, {full_name.title()}! Welcome to {language}.")
````

### Whitespace: invisible characters still count

- `\n` starts a new line
- `\t` inserts a tab
- `.strip()` removes extra whitespace from both ends


````python
menu = "Languages:\n\tPython\n\tJavaScript\n\tC++"
print(menu)

typed_name = "   Alex Chen   "
clean_name = typed_name.strip()

print(f"Before: >{typed_name}<")
print(f"After:  >{clean_name}<")
````

## 6. Numbers: calculate and measure

Python supports familiar arithmetic operators:

| Operation | Symbol | Example |
|---|---:|---:|
| Addition

````python
print(5 + 3)
print(10 - 2)
print(4 * 2)
print(16 / 2)
print(2 ** 3)
````

### Integers and floats

- `int`: a whole number, such as `12`
- `float`: a number with a decimal point, such as `12.5`

If an expression contains a float, the result is usually a float.



````python
items = 3
price_per_item = 2.50
total = items * price_per_item

print(total)
print(type(total))
````

### A tiny floating-point surprise

Computers store many decimal values approximately in binary.



````python
result = 0.2 + 0.1
print(result)
print(round(result, 2))
````
