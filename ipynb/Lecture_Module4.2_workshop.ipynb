## Our Problem-Solving Toolbox

Before typing Python:

1. **Understand** — What should the program actually do?
2. **Identify the data** — What must the program remember?
3. **Find repetition** — What needs to happen more than once?
4. **Find decisions** — Where must the program choose?
5. **Sketch a plan** — Plain English, a diagram, or pseudocode.
6. **Build a small piece** — Run it before adding more.
7. **Test and improve** — Try it, fix it, then continue.

> Pseudocode is simply a plan for humans. It does not need perfect Python syntax.

# Project 1 — Smart Grocery Trip 🛒

You are grocery shopping with a budget. For every item, record its **name, category, and price**. You do not know how many items you will buy. At the end, show the cart, calculate the total, and tell the user whether the trip stayed within budget.

### Think first
- What describes **one item**?
- What structure keeps those related values together?
- What holds **many items**?
- Do we know how many items will be entered?
- Where will the program make decisions?

### Data sketch
`one item → dictionary`  
`many items → list of dictionaries`

### Human-friendly plan
```text
Create an empty cart
Ask for a budget
While shopping continues:
    collect one item
    save it in the cart
    ask whether to continue
Show every item and calculate the total
Use the total to decide what message to show
```

````python
cart = []
budget = float(input("What is your grocery budget? $"))
shopping = True

while shopping:
    name = input("\nItem name: ")
    category = input("Category: ")
    price = float(input("Price: $"))
    cart.append({"name": name, "category": category, "price": price})

    if input("Add another item? (yes/no): ").lower() == "no":
        shopping = False
````

### Pause, run, and inspect
Before adding the next part, test what you have. Does the program remember the data correctly?

````python
print("\n--- YOUR GROCERY CART ---")
total = 0

for item in cart:
    print(f"{item['name']} ({item['category']}): ${item['price']:.2f}")
    total += item["price"]

print(f"Total: ${total:.2f}")
print(f"Budget: ${budget:.2f}")

if total < budget:
    print(f"You are within budget! ${budget-total:.2f} left. 🎉")
elif total == budget:
    print("Exactly on budget. Suspiciously perfect. 😄")
else:
    print(f"You are ${total-budget:.2f} over budget.")
````

### Quick Check
1. Why is one item a dictionary?
2. Why is the cart a list?
3. Why does a `while` loop make sense?
4. What would happen if `cart = []` were inside the loop?
5. Why do we use a `for` loop after shopping is finished?

# Project 2 — Road Trip Planner 🚗

You are planning a road trip with several stops. For each stop, record the **city, miles from the previous stop, and number of nights**. At the end, show the itinerary, total miles, total nights, and classify the trip.

### Work backward from the result
Imagine we want:
```text
Charleston — 210 miles — 2 nights
Savannah — 108 miles — 2 nights
Stops: 2
Total miles: 318
Total nights: 4
Road trip time! 🚗
```

What information must Python remember to create this result?

### Plan
```text
Create an empty list of stops
Repeat:
    collect one stop
    save it
    ask whether there is another stop
For every saved stop:
    display it
    add miles and nights
Classify the trip using total miles
```

````python
stops = []
planning = True

while planning:
    city = input("\nCity: ")
    miles = int(input("Miles from the previous stop: "))
    nights = int(input("Number of nights: "))
    stops.append({"city": city, "miles": miles, "nights": nights})

    if input("Add another stop? (yes/no): ").lower() == "no":
        planning = False

print("\nWhat did Python store?")
print(stops)
````

### Pause, run, and inspect
Before adding the next part, test what you have. Does the program remember the data correctly?

````python
print("\n--- ROAD TRIP ---")
total_miles = 0
total_nights = 0

for stop in stops:
    print(f"{stop['city']} — {stop['miles']} miles — {stop['nights']} nights")
    total_miles += stop["miles"]
    total_nights += stop["nights"]

print(f"Stops: {len(stops)}")
print(f"Total miles: {total_miles}")
print(f"Total nights: {total_nights}")

if total_miles < 200:
    print("Easy drive! 😎")
elif total_miles <= 500:
    print("Road trip time! 🚗")
else:
    print("Big trip — your playlist better be ready. 🎵")
````

### Quick Check
1. What represents one stop? What represents all stops?
2. Why use `while` while collecting data?
3. Why use `for` while summarizing?
4. Where would you add a hotel name?
5. Why classify the trip only after calculating total miles?

# Project 3 — Personal Routine Tracker 🌱

Record several days of everyday routine data. For each day, save the **day, sleep hours, exercise minutes, and mood (`good`, `okay`, or `tired`)**. Then summarize the records and give simple feedback.

### Design questions
- What represents one day's record?
- What stores several days?
- What repeats an unknown number of times?
- What totals or counters are needed?
- What decisions happen after the calculations?

### Short plan
```text
Create an empty list
While there is another day:
    collect one day's information
    save one record
For every record:
    display it
    total sleep and exercise
    count tired days
Calculate average sleep
Use decisions to show feedback
```

````python
records = []
tracking = True

while tracking:
    day = input("\nDay: ")
    sleep = float(input("Hours of sleep: "))
    exercise = int(input("Exercise minutes: "))
    mood = input("Mood (good/okay/tired): ").lower()

    records.append({
        "day": day,
        "sleep": sleep,
        "exercise": exercise,
        "mood": mood
    })

    if input("Record another day? (yes/no): ").lower() == "no":
        tracking = False
````

### Pause, run, and inspect
Before adding the next part, test what you have. Does the program remember the data correctly?

````python
total_sleep = 0
total_exercise = 0
tired_days = 0

print("\n--- ROUTINE SUMMARY ---")
for record in records:
    print(f"{record['day']}: {record['sleep']} hrs sleep, "
          f"{record['exercise']} min exercise, mood: {record['mood']}")
    total_sleep += record["sleep"]
    total_exercise += record["exercise"]

    if record["mood"] == "tired":
        tired_days += 1

average_sleep = total_sleep / len(records)

print(f"Days recorded: {len(records)}")
print(f"Average sleep: {average_sleep:.1f} hours")
print(f"Total exercise: {total_exercise} minutes")
print(f"Tired days: {tired_days}")

if average_sleep >= 8:
    print("Sleep check: Well rested! 😴✨")
elif average_sleep >= 6:
    print("Sleep check: Not bad, but a little more sleep may help.")
else:
    print("Sleep check: Your pillow misses you. 🛏️")

if total_exercise >= 150:
    print("Activity check: Nice active stretch!")
else:
    print("Activity check: There is room for a little more movement.")
````

### Quick Check
1. Which structure represents one day? All days?
2. Why summarize after collecting the records?
3. Where is an `if` inside a `for` loop?
4. Why are sleep and exercise separate decisions?
5. What other everyday information could be added?
