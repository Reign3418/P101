## 1. Your First Function
A function is a named block of code that performs a job.

````python
def greet_user():
    print("Welcome to Python!")

greet_user()
greet_user()
````

## 2. Parameters and Arguments
A **parameter** is a variable in the function definition. An **argument** is the value supplied when we call the function.

````python
def greet_player(username):
    print(f"Welcome, {username}!")

greet_player("Maya")
greet_player("Jordan")
````

### Think Like a Programmer
For every function, ask:
- **What job does it do?**
- **What information does it need?**
- **What should it give back?**

Those questions matter more than memorizing punctuation.

## 3. Multiple Parameters

````python
def show_vehicle_status(vehicle, status):
    print(f"{vehicle}: {status}")

show_vehicle_status("Truck 12", "Ready")
show_vehicle_status("Generator 4", "Maintenance Due")
````

## 4. Positional vs. Keyword Arguments
With positional arguments, order matters. Keyword arguments make the meaning explicit.

````python
def create_character(name, character_class):
    print(f"{name} chose the {character_class} class.")

create_character("Alex", "Mage")
create_character(character_class="Ranger", name="Sam")
````

### Debug It
What is wrong with this call?

```python
create_character("Warrior", "Taylor")
```

Python will run it—but will the **meaning** be correct?

## 5. Default Values
A default is useful when one value is common, but callers can still choose something else.

````python
def start_mission(player, difficulty="Normal"):
    print(f"{player}'s mission difficulty: {difficulty}")

start_mission("Kai")
start_mission("Kai", "Hard")
````

## 6. `return`: Give a Result Back
`print()` displays something. `return` sends a value back to the code that called the function.

````python
def calculate_damage(base_damage, bonus):
    return base_damage + bonus

damage = calculate_damage(30, 12)
print(f"Total damage: {damage}")
````

### Why return instead of only print?
Returned values can be **stored, compared, reused, or passed to another function**.

````python
def check_temperature(temp):
    if temp > 220:
        return "Warning"
    return "Normal"

status = check_temperature(230)
if status == "Warning":
    print("Inspect the cooling system.")
````

## 7. Functions Can Return Different Useful Types

````python
def calculate_readiness(ready, total):
    return ready / total * 100

def player_is_alive(health):
    return health > 0

def make_profile(name, level):
    return {"name": name, "level": level}

print(calculate_readiness(8, 10))
print(player_is_alive(25))
print(make_profile("Nova", 7))
````

## 8. Function Design Practice — Before Coding

### Scenario: Game Health System
We need a program that:
- receives a player's current health and healing amount,
- calculates the new health,
- never allows health to go above 100,
- returns the new health.

Before coding:
1. What is the function's **job**?
2. What **information** does it need?
3. What should it **return**?

````python
def heal_player(health, healing):
    new_health = health + healing
    if new_health > 100:
        new_health = 100
    return new_health

print(heal_player(70, 20))
print(heal_player(95, 20))
````
