## 2. Accessing items: indexes start at 0

Each item has a position called an **index**.

For this list:

```python
planets = ["Mercury", "Venus", "Earth", "Mars"]
```

| Item | Index |
|---|---:|
| Mercury | 0 |
| Venus | 1 |
| Earth | 2 |
| Mars | 3 |

The first item is index `0`, not index `1`.

````python
planets = ["Mercury", "Venus", "Earth", "Mars"]

print(planets[0])
print(planets[2])
````

### Negative indexes

Python can also count backward from the end:

- `-1` → last item
- `-2` → second-to-last item
- `-3` → third-to-last item

````python
print(planets[-1])
print(planets[-2])
````

## 5. Adding items

### `append()` — add to the end

`append()` is one of the most common list operations.

````python
queue = ["Ava", "Noah"]
queue.append("Mia")

print(queue)
````

A common real-world pattern is to begin with an empty list and build it over time.

````python
notifications = []

notifications.append("New message")
notifications.append("Assignment submitted")
notifications.append("System update")

print(notifications)
````

### `insert()` — add at a specific position

`insert(index, value)` places a new item at a chosen index. Existing items shift to make room.

````python
playlist = ["Song A", "Song C"]
playlist.insert(1, "Song B")

print(playlist)
````

## 6. Removing items: three useful tools

Python gives us different removal tools because programs often know different information.

### A. `del` — remove by index

Use `del` when you know **where** the item is and do not need the removed value afterward.

````python
devices = ["laptop", "tablet", "old phone", "watch"]
del devices[2]

print(devices)
````

### B. `pop()` — remove and return an item

`pop()` removes an item **and gives the removed value back**.

With no index, `pop()` removes the last item.

````python
browser_tabs = ["Email", "Docs", "Python Docs"]

closed_tab = browser_tabs.pop()

print("Still open:", browser_tabs)
print("Closed:", closed_tab)
````

````python
players = ["Alex", "Jordan", "Taylor", "Morgan"]

first_player = players.pop(0)

print("Removed:", first_player)
print("Remaining:", players)
````

### C. `remove()` — remove by value

Use `remove(value)` when you know the value rather than its position.

Important: `remove()` deletes only the **first matching value**.

````python
tags = ["python", "beginner", "python", "college"]
tags.remove("python")

print(tags)
````

### Choose the tool

Which would you choose?

Think in terms of **index**, **returned value**, and **known value**.

## 7. Organizing a list

### `sort()` — permanently changes the list

````python
names = ["Zoe", "Aiden", "Maya", "Ben"]

names.sort()
print(names)
````

Use `reverse=True` for descending order.

````python
scores = [88, 100, 73, 95]
scores.sort(reverse=True)

print(scores)
````

### `sorted()` — returns a sorted result without changing the original list

This distinction matters:

- `list.sort()` changes the existing list.
- `sorted(list)` creates a sorted result that you can use while preserving the original order.

````python
arrival_order = ["Maya", "Zoe", "Ben", "Aiden"]

print("Alphabetical:", sorted(arrival_order))
print("Original:    ", arrival_order)
````

### `reverse()` — reverse the current order

`reverse()` does **not** mean alphabetical descending order. It simply flips the current sequence.

````python
steps = ["start", "process", "finish"]
steps.reverse()

print(steps)
````

### Common surprise: methods that change a list often return `None`

Try this:

````python
numbers = [3, 1, 2]
result = numbers.sort()

print("numbers:", numbers)
print("result:", result)
````

`sort()` changes `numbers` in place. It is not designed to produce a new sorted list.

If you need a separate sorted result, use:

```python
result = sorted(numbers)
```

## 8. Finding the length with `len()`

`len()` returns the number of items in a list.

````python
courses = ["Python", "Networking", "Cybersecurity"]

print(len(courses))
````

### Index vs. length

If a list has 4 items:

- its length is `4`
- its valid positive indexes are `0`, `1`, `2`, and `3`

That difference is one reason beginners often encounter an `IndexError`.

Run the next cell intentionally. Errors are part of programming.

````python
pets = ["cat", "dog", "rabbit"]

# Uncomment the next line to intentionally create an IndexError.
# print(pets[3])
````

When you see:

`IndexError: list index out of range`

ask:

1. What is the length of my list?
2. Which index did I request?
3. Is that index valid?
4. Did the list change earlier in the program?

A quick debugging habit:

````python
pets = ["cat", "dog", "rabbit"]

print("List:", pets)
print("Length:", len(pets))
print("Last valid positive index:", len(pets) - 1)
````

### Edge case: an empty list

Negative indexing is convenient, but `empty_list[-1]` still fails because there is no item to return.

````python
empty_list = []
print("Length:", len(empty_list))

# Uncomment to test:
# print(empty_list[-1])
````
