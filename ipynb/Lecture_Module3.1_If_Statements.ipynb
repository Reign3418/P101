## 1. Conditions and Boolean Values
A **condition** is an expression Python can evaluate as `True` or `False`.

Predict before running:

````python
print(5 > 3)
print(10 == 10)
print(7 < 2)
print("cat" == "dog")
````

### `=` is not `==`
```python
score = 90
```
means **assign 90 to `score`**.

```python
score == 90
```
means **ask whether `score` equals 90**.

> `=` → put a value somewhere  
> `==` → ask a question

````python
score = 90
print(score == 90)
print(score == 100)
````

## 2. Comparing Values

| Operator | Meaning |
|---|---|
| `==` | equal |
| `!=` | not equal |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal |
| `<=` | less than or equal |

````python
age = 21
print(age == 21)
print(age != 18)
print(age >= 18)
print(age < 30)
````

### Strings and case
String comparisons are case-sensitive. Normalize case when capitalization should not matter.

````python
answer = "YES"
print(answer == "yes")
print(answer.lower() == "yes")
````

## 3. Checking Multiple Conditions

### `and`
**Both** conditions must be `True`.

### `or`
**At least one** condition must be `True`.

````python
age = 22
has_id = True
print(age >= 18 and has_id)

is_weekend = False
is_holiday = True
print(is_weekend or is_holiday)
````

## 4. Membership: `in` and `not in`
Lists become more useful when we can ask whether something is inside them.

````python
available_drinks = ["coffee", "tea", "water"]
print("tea" in available_drinks)
print("soda" in available_drinks)
print("soda" not in available_drinks)
````

## 5. A Simple `if` Statement
An `if` block runs **only when its condition is True**.

```python
if condition:
    # do something
```

Remember the colon and indentation.

````python
battery = 12
if battery < 20:
    print("Low battery. Please charge soon.")
print("Battery check complete.")
````

## 6. `if-else`: Two Possible Paths
Use `else` when exactly one of two paths should run.

````python
password = "python123"
if password == "python123":
    print("Access granted.")
else:
    print("Access denied.")
````

The `else` block needs no condition. It means:

> If the first condition is false, do this instead.

## 7. `if-elif-else`: More Than Two Choices
Use `elif` when you have several mutually exclusive choices. Python selects the **first matching branch**.

````python
temperature = 72

if temperature >= 90:
    print("Hot")
elif temperature >= 70:
    print("Warm")
elif temperature >= 50:
    print("Cool")
else:
    print("Cold")
````

### Order matters
Python checks top to bottom and stops after the first match.

You also do **not** always need an `else`. Use it only when a catch-all case makes sense.

## 8. `elif` Chain vs Independent `if` Statements

**Need one matching choice?** Use an `if-elif-else` chain.

**Need every true condition to act?** Use independent `if` statements.

````python
features = ["dark mode", "notifications"]

if "dark mode" in features:
    print("Enable dark mode.")

if "notifications" in features:
    print("Enable notifications.")
````

## 9. Combining `if` with Lists and Loops
Now the pieces start working together.

````python
orders = ["coffee", "tea", "sold out", "smoothie"]

for order in orders:
    if order == "sold out":
        print("Sorry, choose something else.")
    else:
        print(f"Preparing {order}...")
````

### Checking for an empty list
An empty list is false in a Boolean context, so this is common:

```python
if messages:
    ...
```

````python
messages = []

if messages:
    print("You have messages.")
else:
    print("Inbox zero! Enjoy it while it lasts. 😄")
````

### Checking requested items against available items

````python
available = ["coffee", "tea", "water"]
requested = ["tea", "soda"]

for drink in requested:
    if drink in available:
        print(f"{drink.title()} is available.")
    else:
        print(f"Sorry, no {drink}.")
````

## 10. Modern Python Notes
The core `if` syntax remains standard modern Python.

Useful habits:
- prefer clear conditions
- `if logged_in:` is usually clearer than `if logged_in == True:`
- use `in` for membership
- use `elif` when one branch should win
- use independent `if`s when multiple actions may apply

Python also supports chained comparisons:

```python
18 <= age < 65
```

````python
age = 28
print(18 <= age < 65)
````
