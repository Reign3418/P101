## 0. Warm-up — Data with Labels

Lists are great when position matters. Dictionaries are great when values need **labels**.

```python
student = {"name": "Maya", "major": "Cybersecurity", "credits": 42}
```

Instead of asking “What is item 0?”, we can ask “What is the student's **name**?”

A dictionary is a collection of **key-value pairs**.

## 1. Creating a Dictionary

Dictionary syntax uses curly braces. Each key is connected to a value with a colon.

```python
profile = {
    "username": "maya22",
    "active": True,
    "login_count": 17
}
```

Keys act like labels. Values can be strings, numbers, Booleans, lists, or even other dictionaries.

````python
profile = {"username": "maya22", "active": True, "login_count": 17}
print(profile)
print(type(profile))
````

## 2. Accessing Values

Use a key inside square brackets:

```python
dictionary["key"]
```

This is different from list indexing:

```python
colors[0]            # position
profile["username"]  # label
```

````python
profile = {"username": "maya22", "active": True, "login_count": 17}
print(profile["username"])
print(profile["login_count"])
````

## 3. Adding and Updating

Assign to a **new key** to add data. Assign to an **existing key** to update it.

````python
profile = {"username": "maya22"}
profile["active"] = True
profile["login_count"] = 1
print(profile)

profile["login_count"] = 2
print(profile)
````

### Starting Empty

An empty dictionary is useful when your program learns information gradually.

````python
settings = {}
settings["theme"] = "dark"
settings["notifications"] = True
settings["font_size"] = 14
print(settings)
````

## 4. Removing Data

Use `del` when you want a key-value pair gone. Use `.pop()` when you also want the removed value back.

````python
settings = {"theme": "dark", "notifications": True, "font_size": 14}
del settings["font_size"]
old_theme = settings.pop("theme")
print(old_theme)
print(settings)
````

## 5. Missing Keys and `.get()`

A square-bracket lookup for a missing key raises `KeyError`.

When missing data is normal, `.get()` can provide a fallback:

```python
dictionary.get(key, default)
```

Use `[]` when a key **must exist**. Use `.get()` when a missing key is an expected possibility.

````python
user = {"name": "Noah", "city": "Raleigh"}
print(user.get("city", "Unknown"))
print(user.get("phone", "Not provided"))

# Uncomment to compare:
# print(user["phone"])
````

## 6. Looping Through Dictionaries

Three common patterns:

| Need | Pattern |
|---|---|
| keys + values | `for k, v in d.items()` |
| keys | `for k in d` or `d.keys()` |
| values | `for v in d.values()` |

````python
course = {"code": "CSC-121", "room": "210", "credits": 3}

print("KEY + VALUE")
for key, value in course.items():
    print(key, value)

print("\nKEYS")
for key in course:
    print(key)

print("\nVALUES")
for value in course.values():
    print(value)
````

## 7. Order and Unique Keys

In current Python, dictionaries preserve **insertion order**. Older tutorials may describe dictionaries as unordered.

But dictionaries are still designed for **lookup by key**, not numeric position.

Keys are unique. Reusing a key replaces its earlier value.

````python
person = {"name": "Maya", "name": "Ava"}
print(person)

person = {"name": "Maya", "hobbies": ["gaming", "hiking", "music"]}
print(person["hobbies"])
````

## 8. A List of Dictionaries

One dictionary can represent one record; a list can hold many records. This pattern is common in applications and JSON/API data.

````python
products = [
    {"name": "Notebook", "price": 4.99, "in_stock": True},
    {"name": "Headphones", "price": 29.99, "in_stock": False},
    {"name": "Water Bottle", "price": 12.50, "in_stock": True}
]

for product in products:
    if product["in_stock"]:
        print(f'{product["name"]}: available')
    else:
        print(f'{product["name"]}: sold out')
````

## 9. A List Inside a Dictionary

A dictionary value can contain several related values in a list.

````python
playlist = {
    "name": "Study Mix",
    "genres": ["lo-fi", "instrumental", "jazz"],
    "songs": 24
}

for genre in playlist["genres"]:
    print(genre)
````

## 10. A Dictionary Inside a Dictionary

A value can also be another dictionary.

Read nested access **from the outside in**:

```python
campus["library"]["floor"]
```

Campus → library → floor.

If the code starts looking like punctuation soup, inspect one level at a time. 🍜

````python
campus = {
    "library": {"floor": 2, "open": True},
    "lab": {"floor": 1, "open": False}
}

print(campus["library"]["floor"])
print(campus["lab"]["open"])
````

## 11. Modern Python Notes

The fundamentals above are current Python. A few useful habits:

- dictionaries preserve insertion order
- use `.get()` when missing keys are expected
- use clear, consistent key names
- use `.items()` when you need both key and value
- avoid unnecessary deep nesting

Modern Python also supports dictionary merging with `|`:

````python
defaults = {"theme": "light", "font_size": 14}
choices = {"theme": "dark"}
settings = defaults | choices
print(settings)
````
