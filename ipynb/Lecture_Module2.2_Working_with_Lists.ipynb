## 1. The `for` Loop
A `for` loop visits items one at a time:

```python
for item in collection:
    # do something with item
```

Read it as: **For each item in the collection, do this.**

Use a meaningful singular loop variable when possible: `city` in `cities`, `file` in `files`, `score` in `scores`.

````python
cities = ["Raleigh", "Chicago", "Seattle"]
for city in cities:
    print(f"I would like to visit {city}.")
````

## 2. Indentation: Python Cares About Space
The indented block belongs to the loop:

```python
for file in files:
    print(file)
    print("Checked!")
```

Both lines repeat. A line back at the left margin runs **after** the loop.

Python uses indentation as syntax, not decoration.

````python
files = ["report.pdf", "notes.txt", "photo.jpg"]
for file in files:
    print(f"Checking {file}")
    print("Done with this file.")
print("All files checked!")
````

## 3. `range()` — Let Python Generate the Numbers
`range()` is useful for integer sequences:

```python
range(start, stop, step)
```

**The stop value is not included.**

- `range(5)` → 0, 1, 2, 3, 4
- `range(1, 6)` → 1, 2, 3, 4, 5
- `range(0, 10, 2)` → 0, 2, 4, 6, 8

A `range` is an iterable object. Use `list(range(...))` only when you actually need a list.

````python
for number in range(1, 6):
    print(number)

numbers = list(range(1, 6))
print(numbers)
````

## 4. Numerical Lists
Useful built-ins:
- `min()` → smallest
- `max()` → largest
- `sum()` → total

````python
daily_steps = [6420, 8100, 7350, 10200, 5900]
print("Lowest:", min(daily_steps))
print("Highest:", max(daily_steps))
print("Total:", sum(daily_steps))
````

### Building a numerical list
A loop can transform values and build a new list.

````python
squares = []
for number in range(1, 6):
    squares.append(number ** 2)
print(squares)
````

## 5. List Comprehensions — A Compact Pattern
Longer version:

```python
squares = []
for number in range(1, 6):
    squares.append(number ** 2)
```

Compact version:

```python
squares = [number ** 2 for number in range(1, 6)]
```

Both work. Use a comprehension when it is clear. There is no prize for squeezing your whole program onto one line. 😄

````python
squares = [number ** 2 for number in range(1, 6)]
print(squares)
````

## 6. Slicing — Work with Part of a List
Indexing gives **one item**. Slicing gives a **group of items**:

```python
movies[start:stop]
```

The start is included; the stop is excluded.

````python
movies = ["Dune", "Arrival", "Shrek", "Barbie", "Interstellar"]
print(movies[1:4])
print(movies[:3])
print(movies[2:])
print(movies[-2:])
````

### Slice mental model
- `movies[1:4]` → indexes 1, 2, 3
- `movies[:3]` → first three
- `movies[2:]` → index 2 through the end
- `movies[-2:]` → last two

> **Start here, stop before there.**

## 7. Looping Through a Slice
A slice is a list, so you can loop through it.

````python
menu = ["salad", "pasta", "tacos", "curry", "soup"]
print("Today's featured items:")
for item in menu[:3]:
    print(f"- {item}")
````

## 8. Copying a List: `.copy()` vs `=`
This is subtle:

```python
friend_movies = my_movies
```

does **not** make an independent list. Both names refer to the same list.

For an independent shallow copy:

```python
friend_movies = my_movies.copy()
```

A full slice `my_movies[:]` also creates a shallow copy.

````python
my_movies = ["Dune", "Arrival", "Shrek"]
friend_movies = my_movies
friend_movies.append("Barbie")
print("my_movies:    ", my_movies)
print("friend_movies:", friend_movies)
````

````python
my_movies = ["Dune", "Arrival", "Shrek"]
friend_movies = my_movies.copy()
friend_movies.append("Barbie")
print("my_movies:    ", my_movies)
print("friend_movies:", friend_movies)
````

For simple lists, remember:

- `b = a` → another name for the same list
- `b = a.copy()` → a new outer list with the same current items

We will save deeper shallow-copy questions involving nested objects for later.

## 9. Tuples — A Collection That Should Not Change
A tuple is an **immutable sequence**.

```python
screen_size = (1920, 1080)
```

You can index and loop through it, but you cannot replace an individual item.

````python
screen_size = (1920, 1080)
print(screen_size[0])
for dimension in screen_size:
    print(dimension)

# Uncomment to intentionally create a TypeError:
# screen_size[0] = 1280
````

### Modern tuple detail
The comma defines a tuple. A one-item tuple needs a trailing comma:

```python
one_item = (42,)
not_a_tuple = (42)
```

````python
one_item = (42,)
not_a_tuple = (42)
print(type(one_item))
print(type(not_a_tuple))
````

## 10. Modern Python Note
The core concepts here remain current Python.

A useful modern habit: loop over items directly when you only need the items.

```python
for name in names:
    print(name)
```

If you genuinely need both a position and an item, `enumerate()` is often clearer than manually managing indexes:

```python
for position, name in enumerate(names, start=1):
    print(position, name)
```

You do **not** need to master `enumerate()` today.

Also, `range()` is an iterable range object; it does not pre-build a list unless you call `list()`.

````python
names = ["Maya", "Leo", "Ava"]
for position, name in enumerate(names, start=1):
    print(f"{position}. {name}")
````
