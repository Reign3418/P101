## 1. Quick Review

````python
def calculate_score(points, bonus=0):
    return points + bonus

print(calculate_score(100))
print(calculate_score(100, 25))
````

## 2. Optional Information with `None`
Sometimes information truly may be missing. `None` is Python's way to represent “no value here.”

````python
def build_player(name, nickname=None):
    if nickname:
        return f"{name} ({nickname})"
    return name

print(build_player("Jordan"))
print(build_player("Jordan", "Shadow"))
````

## 3. Functions + `while` Loops
A loop can keep collecting information; a function can process each piece.

````python
def classify_score(score):
    if score >= 90:
        return "Excellent"
    elif score >= 70:
        return "Good"
    return "Keep practicing"

scores = [95, 76, 61]
for score in scores:
    print(score, classify_score(score))
````

### Connecting to last module
A `while` loop is especially useful when we do **not know in advance** how many times the user will repeat an action.

````python
def create_report(name, status):
    return f"{name}: {status}"

while True:
    name = input("Equipment name (or 'quit'): ")
    if name.lower() == "quit":
        break

    status = input("Status: ")
    print(create_report(name, status))
````

## 4. Passing a List to a Function

````python
def show_players(players):
    for player in players:
        print(f"- {player}")

team = ["Avery", "Chris", "Morgan"]
show_players(team)
````

## 5. A Function Can Modify a List

````python
def complete_tasks(todo, completed):
    while todo:
        task = todo.pop()
        completed.append(task)

todo = ["check logs", "update software", "run test"]
completed = []

complete_tasks(todo, completed)

print("Todo:", todo)
print("Completed:", completed)
````

### Protecting the Original List
If you do **not** want the function to change the original list, pass a copy:

```python
complete_tasks(todo[:], completed)
```

Use this intentionally—not automatically. Sometimes modifying the original is exactly what the program needs.

## 6. `*args`: An Unknown Number of Positional Arguments
The `*` collects extra positional arguments into a tuple.

````python
def total_points(*scores):
    return sum(scores)

print(total_points(10, 20))
print(total_points(10, 20, 30, 40))
````

## 7. `**kwargs`: An Unknown Number of Keyword Arguments
The `**` collects extra named arguments into a dictionary.

````python
def build_game_item(name, **details):
    item = {"name": name}
    item.update(details)
    return item

sword = build_game_item("Iron Sword", damage=25, rarity="Common")
print(sword)
````

### Keep Perspective
You do **not** need `*args` or `**kwargs` for every function. Use them when the number of inputs is genuinely flexible.

## 8. Modules: Put Functions in Another File
As programs grow, functions do not all need to live in one file.

````python
# Conceptual example:
# game_tools.py
# def calculate_damage(...):
#     ...

# main.py
# import game_tools
# game_tools.calculate_damage(...)
````

Common import styles:

```python
import module_name
from module_name import function_name
import module_name as short_name
```

For now, prefer clear imports. Avoid `from module_name import *` because it can make names harder to track.

## 9. Program Design: Small Functions Working Together

### Mini Diagnostic / Status System

Requirement:
1. Check a numeric reading.
2. Create a readable status message.
3. Process several readings.
4. Count warnings.

Before coding, identify the smaller jobs.

````python
def check_reading(value, limit):
    if value > limit:
        return "WARNING"
    return "OK"

def format_result(name, value, status):
    return f"{name}: {value} -> {status}"

def analyze_readings(readings, limit):
    warnings = 0

    for name, value in readings.items():
        status = check_reading(value, limit)
        print(format_result(name, value, status))

        if status == "WARNING":
            warnings += 1

    return warnings

sensor_data = {
    "Sensor A": 72,
    "Sensor B": 104,
    "Sensor C": 89
}

warning_count = analyze_readings(sensor_data, 100)
print(f"Warnings: {warning_count}")
````

### Notice the structure

`check_reading()` → one decision  
`format_result()` → one formatting job  
`analyze_readings()` → coordinates the process

No function has to understand the entire program.
